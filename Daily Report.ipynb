```python
def generate_report(time_log, morning_study, training, afternoon_study, challenges, plan_tomorrow, additional_notes='None'):

    report = '**Daily Activity Report**\n'
    report += datetime.today().strftime('%m/%d/%y') + '\n'
    report += 'Name: Mikael Huff\n'
    report += 'Technology: Data Science (DS)\n'
    report += 'CDM: Divesh Gupta\n\n'

    report += '🕒 **Time Log & Activities**\n' +\
                '08:00 - 09:00: Self-Study\n' +\
                '09:00 - 13:00: Training\n' +\
                '13:00 - 14:00: Lunch Break\n' +\
                '14:00 - 18:00: Training Session\n\n'
    
    report += '✅ **Summary of Work Done**\n'
    if morning_study != '':
        report += f'Morning Self-study:\n{morning_study}\n\n'
    report += f'Training:\n{training}\n\n'
    report += f'Afternoon Self-study:\n{afternoon_study}\n\n'

    report +='🚧 **Issues/Challenges Faced**'
    report += f'\n{challenges}\n\n'

    if plan_tomorrow != '':
        report += '📅 **Plan for Tomorrow**\n'
        report += f'{plan_tomorrow}\n'

    report += '📝 **Additional Notes**'
    report += f'\n{additional_notes}\n'

    return report
```
