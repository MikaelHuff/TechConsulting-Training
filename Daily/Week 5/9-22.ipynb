```python
class Solution:
    def __init__(self):
        self.N = 0

    def constructTransformedArray(self, nums: list[int]) -> list[int]:
        self.N = len(nums)
        result = [0] * self.N
        for i in range(self.N):
            
            if nums[i] > 0:
                result[i] = nums[self.overflow(i + nums[i])]
            elif nums[i] < 0:
                # continue
                result[i] = nums[self.overflow(i + nums[i])]
            else:
                result[i] = nums[i]
            # break
        return result

    
    def overflow(self, val):
        print(self.N, val)
        return val % self.N
        if val >= self.N:
            return self.overflow(val - self.N)
        elif val < 0:
            return self.overflow(self.N + val)
        return val
    
s = Solution()
s.constructTransformedArray([-1,4,-1])
```

```python
import numpy as np
```

```python
random_vals = (np.random.rand(1001,2)-.5)*np.random.randint(low=0,high=10)
m_true,b_true = random_vals[0,:]
x_train,y_train = random_vals[1:801,:].transpose()
x_test, y_test = random_vals[802:,:].transpose()

```

```python
x_bar = np.mean(x_train)
y_bar = np.mean(y_train)

```

```python
m_pred = np.dot(x_bar-x_train,y_bar-y_train)/np.sum((x_train-x_bar)**2)
b_pred =  y_bar - x_bar * m_pred
```

```python

y_pred = m_pred * x_test + b_pred
err = np.mean((y_test-y_pred)**2)
print(y_test.shape,y_pred.shape)
print(err)
```

```python

print('             True  |  Pred  ')
print(f'Slope:     {m_true:.4f} | {m_pred:.4f}')
print(f'Intercept: {b_true:.4f} | {b_pred:.4f}')
print('-'*30)
print(f'MSE: {err:.4f}')
```
