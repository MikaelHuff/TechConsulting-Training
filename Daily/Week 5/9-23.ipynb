```python
class Solution:
    def compareVersion(self, version1: str, version2: str) -> int:
        v1_split = version1.split('.')
        v2_split = version2.split('.')
        v1_N = len(v1_split)
        v2_N = len(v2_split)
        v1_split += ['0'] * (v2_N - v1_N)
        v2_split += ['0'] * (v1_N - v2_N)
        for i in range(max(v1_N,v2_N )):
            print(v1_split[i], v2_split[i])
            if int(v1_split[i]) < int(v2_split[i]):
                return -1
            elif int(v1_split[i]) > int(v2_split[i]):
                return 1
        return 0
s = Solution()
print(s.compareVersion('1', '1.1'))
```

```python
class Solution:
    def __init__(self):
        self.nums = {}

    def getKth(self, lo: int, hi: int, k: int) -> int:
        self.nums = {'1': 0}
        results = []
        for val in range(lo,hi+1):
            results.append(self.check_num(val))
            # break
        print(self.nums)
        print(results)
        results_sorted = sorted(list(range(lo,hi+1)), key=lambda x: self.nums[str(int(x))])
        print(results_sorted)
        return results_sorted[k-1]

    def check_num(self, val):
        val_str = str(int(val))
        # print(val)
        if val_str in self.nums:
            return self.nums[val_str]
        elif val % 2 == 0:
            count = self.check_num(val/2) + 1
        else:
            count = self.check_num(3*val + 1) + 1
        self.nums[val_str] = count
        
        return count
        

b = Solution()
b1 = [12,15,2]
b.getKth(*b1)
# list(range(5,10))
```
