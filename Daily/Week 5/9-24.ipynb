```python
my_list=[1,2,6,4,3,8,9,22,8]
def example_func(input_list):
    list_sorted = sorted(input_list)
    odds = [val for val in list_sorted if val % 2]
    evens = [val for val in list_sorted if not val % 2]
    return {'odds':odds, 'evens':evens[::-1]}
print(example_func(my_list))
```

```python
def find_primes(num):
    result = []
    def is_prime(_num):
        for i in range(2, _num):
            if _num % i == 0:
                return False
        return True
    for i in range(2, num+1):
        if num % i == 0 and is_prime(i):
            result.append(i)

    return result

print(find_primes(1257))

```
