```python
def reverse_str(text):
    return text[::-1]


def check_palindrome(text):
    return text.lower() == reverse_str(text.lower())

name = 'racecar'
print(reverse_str(name))
print(check_palindrome(name))
```

```python
from math import prod, lcm, gcd
class Solution:
    def maxLength(self, nums: list[int]) -> int:
        max_size = 0
        sub_array = []
        left = 0
        right = 1
        while left+max_size < len(nums):
            for right in range(max_size, len(nums)+1):
                sub_array = nums[left:right]                
                # print(sub_array)
                if prod(sub_array) == lcm(*sub_array)*gcd(*sub_array):
                    if len(sub_array) > max_size:
                        max_size = len(sub_array)
            left += 1
        return max_size
        

nums = [1,2,1,2,1,1,1]
s = Solution()
print(s.maxLength(nums))
```
