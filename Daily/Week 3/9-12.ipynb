```python
class Solution:
    def canBeIncreasing(self, nums: list[int]) -> bool:
        count = 1
        for i in range(1,len(nums)):
            
            print(nums[i],nums[i-1], nums)
            if nums[i] <= nums[i-1]:
                if count == 1:
                    count = 0
                    if i >= 2 and nums[i] <= nums[i-2]:
                        nums[i] = nums[i-1]
                else:
                    return False
        return True
```

```python
def doesAliceWin(s: str) -> bool:
    vowel_am = 0
    for vowel in ['a','e','i','o','u']:
        vowel_am += s.count(vowel)
    return (vowel_am % 2) == 0
```

```python
s = "leetcoder"

print(doesAliceWin(s))
```
