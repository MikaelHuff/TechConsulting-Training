```python
def maxSubstrings(word: str) -> int:
    return find_word(word, 0)
        


def find_word(word:str, depth):
    if len(word) < 4:
        return depth
    print('\nword: ',word, depth)
    max_str = 0
    for i in range(3, len(word)):
        if word[0] == word[i]:
            print('1', word[0:i])
            print('2', word[i+1:])
            max_str = max(max_str, find_word(word[i+1:], depth+1))
    max_str = max(max_str, find_word(word[1:], depth))

    return max(max_str, depth)


word = "ccaceddad"
print('result: ', maxSubstrings(word))
```

```python
import time
import numpy as np
import random


def dict_check(test):
    dict_test = dict.fromkeys(test, None)
    for i in range(len(test)+1):
        if i not in dict_test:
            return i

def in_check(test):
    num = len(test)+1
    for i in range(num):
        if i not in test:
            return i

def sum_check(test):
    num = len(test)+1
    sum_exp = num * (num+1)/2
    sum_act = sum(test)
    return sum_exp-sum_act
    
def np_check(test):

    num = test.size
    # num= len(test)
    sum_exp = num * (num+1)/2
    sum_act = np.sum(test)
    return sum_exp-sum_act


def monte_carlo(runs, func):
    start = time.time()
    for i in range(runs):
        random.seed(i)
        n = random.randint(0, 10000)
        test = list(range(n))
        random.shuffle(test)
        truth = test[0]
        result = func(np.array(test[1:]))
        assert(truth==result, 'ans: %f and result: %f',truth,result)
    end = time.time()
    return (end-start)


runs = 300
print('dictionary: ', monte_carlo(runs, dict_check))
# print('check list: ', monte_carlo(runs, in_check))
print('sum: ', monte_carlo(runs, sum_check))
print('np: ', monte_carlo(runs, np_check))
```

```python
class Solution:
    def minEatingSpeed(piles: list[int], h: int) -> int:
        for k in range(1, max(piles)+1):
            if k == 30:
                print(k)            
            total_bananas = 0
            for i in range(len(piles)):
                total_bananas += piles[i] + k-(piles[i]-1) % k - 1
                # print(total_bananas, piles[i] + k-(piles[i]-1) % k - 1)
            if total_bananas / h <= k:
                return k
            

s = Solution
test = [30,11,23,4,20]
h = 5
print(s.minEatingSpeed(test,h))
```
