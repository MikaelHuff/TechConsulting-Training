```python
import cv2
import os

def make_video(image_folder, output_video_name):
    # Define video parameters
    fps = 20
    fourcc = cv2.VideoWriter_fourcc(*'MP4V')  # Or 'MJPG' for .avi

    # Get and sort image files
    # print(image_folder)
    images = [img for img in os.listdir(image_folder) if img.endswith((".png", ".jpg", ".jpeg"))]
    # print(len(images))
    images.sort()

    # Get dimensions from the first image
    first_image_path = os.path.join(image_folder, images[0])
    frame = cv2.imread(first_image_path)
    height, width, layers = frame.shape
    frame_size = (width, height)

    # Initialize VideoWriter
    video_writer = cv2.VideoWriter(output_video_name, fourcc, fps, frame_size)

    # Write images to video
    for image_name in images:
        image_path = os.path.join(image_folder, image_name)
        frame = cv2.imread(image_path)
        video_writer.write(frame)

    # Release the VideoWriter
    video_writer.release()

    print(f"Video '{output_video_name}' created successfully!")
```

```python
main_dir = os.path.join(os.getcwd(), 'Lego_Tracking\\Test\\')
output_dir = os.path.join(os.getcwd(), 'videos/')
# print('0',  os.getcwd())
# print('1', main_dir)
# print('2', output_dir)
if os.path.exists(output_dir) == False:
    os.mkdir(output_dir)
    
for dir_name in os.listdir(main_dir):
    print('3', dir_name)
    image_dir = os.path.join(main_dir,dir_name+'/img1/')
    # print('4', os.listdir(image_dir))
    make_video(image_dir, output_dir +dir_name + '-test.mp4')
```
